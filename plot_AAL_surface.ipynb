```python
import scipy.io
import numpy as np
import pyvista as pv
import matplotlib.pyplot as plt
import modules.lib_aal_plot as aplt

import sys
if aplt:
    del sys.modules['modules.lib_aal_plot']
    del aplt
    import modules.lib_aal_plot as aplt

pv.set_jupyter_backend('static')
aal = scipy.io.loadmat('modules/aal_cortex_map_olf294_fix_v7.mat', squeeze_me=True)

aal.keys()
```

```python

import sys
if aplt:
    del sys.modules['modules.lib_aal_plot']
    del aplt
    import modules.lib_aal_plot as aplt

N           = 306
labels      = aal['comm']['AAL306'].item()
tick_pos    = np.append(0,np.unique(labels))
tick_labels = np.append('',aplt.get_node_initials(aal['labels']['AAL09comm']))
fig,ax,cbar = aplt.plot_trisurf_wrapper(
    node_values=labels,
    aal=aal,
    cmap='tab20',
    discrete=True,
    clim=np.max(labels),
    brain_background_color=(0,0,0),
    panel_w_in=2.5,     # inches
    dpi=300,
    show_subcortical_nodes=True,
    show_subcortical_symbol_leg=True,
    add_colorbar=True,
    colorbar_kwargs=dict(
        tick_positions=tick_pos,      # WHERE
        tick_labels=tick_labels,  # WHAT
        cbar_text='RSN',
        width=0.35,
        height=0.04,
        y=0.06,
        labelsize=10,
        fontsize=12
    ),
)

plt.show()
```
